## Replacing Pure Geometric with Visual Semantics

### Old Approach
Given views `i`, `j` and its mask:
1. Find observers
    = views: at leat 30 % of points of both masks are visible from that view (seperately, for each mask)
2. Evaluate support
* For each of the observes: 
    * go through all the detected masks (`k`) in that frame and check if:
    * 80 % of mask `i` (visible points) is inside `k`'s point cloud?
    * 80 % of mask `j` (visible points) is inside `k`'s point cloud
* If true -> is supported

3. Aggregate into concencus rate: $c(i, j) = \frac{n_{supporter}}{n}$
4. Turn into graph edge: hard threshold $c(i, j) \geq 0.9$

### New Approach
-> Replacing `Step 2`
* Given the view (both of the masks are visible): is there some object here representing them both as one thing?
1. Pass view as it is 
2. Pass segmented masks in the view
3. Pass segmented mask one by one
